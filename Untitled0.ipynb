```python

from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np

# Cargar el archivo CSV
file_path = '/content/drive/MyDrive/titanic.csv'
df = pd.read_csv(file_path)


print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas.")
df.head()
```

```python

total_pasajeros = len(df)
passenger_ids_unicos = df['PassengerId'].nunique()

print(f"Total de filas (pasajeros) en el dataset: {total_pasajeros}")
print(f"Cantidad de PassengerId distintos: {passenger_ids_unicos}")

if total_pasajeros == passenger_ids_unicos:
    print("Resultado: Sí coinciden ambos valores.")
else:
    print("Resultado: No coinciden ambos valores.")
```

```python

pasajeros_por_clase = df['Pclass'].value_counts().sort_index().reset_index()
pasajeros_por_clase.columns = ['Clase (Pclass)', 'Cantidad de Pasajeros']


pasajeros_por_clase['Clase_Nombre'] = pasajeros_por_clase['Clase (Pclass)'].map({1: 'Primera', 2: 'Segunda', 3: 'Tercera'})

display(pasajeros_por_clase)

clase_max = pasajeros_por_clase.loc[pasajeros_por_clase['Cantidad de Pasajeros'].idxmax()]
print(f"\nLa clase con más pasajeros registrados es la Clase {clase_max['Clase (Pclass)']} ({clase_max['Clase_Nombre']}) con {clase_max['Cantidad de Pasajeros']} pasajeros.")
```

```python

puertos_conteo = df['Embarked'].value_counts(dropna=False).reset_index()
puertos_conteo.columns = ['Puerto', 'Cantidad de Pasajeros']


puertos_conteo['Puerto_Descripcion'] = puertos_conteo['Puerto'].map({
    'C': 'Cherburgo',
    'Q': 'Queenstown',
    'S': 'Southampton',
    np.nan: 'Sin registrar'
})

display(puertos_conteo)


nulos_embarked = df['Embarked'].isna().sum()
print(f"\nNúmero de pasajeros con el puerto sin registrar (valores faltantes): {nulos_embarked}")
```

```python

tarifas_por_clase = df.groupby('Pclass')['Fare'].agg(
    Tarifa_Promedio='mean',
    Tarifa_Minima='min',
    Tarifa_Maxima='max'
).reset_index()

tarifas_por_clase['Clase_Nombre'] = tarifas_por_clase['Pclass'].map({1: 'Primera', 2: 'Segunda', 3: 'Tercera'})
display(tarifas_por_clase)
```

```python


df['Embarked_Clean'] = df['Embarked'].fillna('Sin registrar')
tabla_clase_puerto = pd.crosstab(df['Pclass'], df['Embarked_Clean'], margins=True)

display(tabla_clase_puerto)
```

```python

df['Survived_Label'] = df['Survived'].map({0: 'Falleció', 1: 'Sobrevivió'})
supervivencia_clase = pd.crosstab(df['Pclass'], df['Survived_Label'], margins=True)

display(supervivencia_clase)
```

```python

df['FamilySize'] = df['SibSp'] + df['Parch']

def clasificar_familia(size):
    if size == 0:
        return '0 familiares (Viajaba solo)'
    elif size == 1:
        return '1 familiar'
    else:
        return '2 o más familiares'

df['Grupo_Familia'] = df['FamilySize'].apply(clasificar_familia)
familiares_conteo = df['Grupo_Familia'].value_counts().reset_index()
familiares_conteo.columns = ['Categoría de Familiares', 'Cantidad de Pasajeros']

display(familiares_conteo)
```

```python

def clasificar_edad(edad):
    if pd.isna(edad):
        return 'Sin registrar'
    elif edad < 18:
        return 'Menor de 18 años'
    else:
        return '18 años o más'

df['Rango_Edad'] = df['Age'].apply(clasificar_edad)


edad_por_clase = pd.crosstab(df['Pclass'], df['Rango_Edad'], margins=True)
display(edad_por_clase)

print(f"Total de valores faltantes en la columna Age: {df['Age'].isna().sum()}")
```

```python

def clasificar_edad(edad):
    if pd.isna(edad):
        return 'Sin registrar'
    elif edad < 18:
        return 'Menor de 18 años'
    else:
        return '18 años o más'

df['Rango_Edad'] = df['Age'].apply(clasificar_edad)


edad_por_clase = pd.crosstab(df['Pclass'], df['Rango_Edad'], margins=True)
display(edad_por_clase)

print(f"Total de valores faltantes en la columna Age: {df['Age'].isna().sum()}")
```

```python

total_tickets_distintos = df['Ticket'].nunique()
conteo_tickets = df['Ticket'].value_counts()


tickets_multiples = conteo_tickets[conteo_tickets > 1]
num_tickets_multiples = len(tickets_multiples)

print(f"Cantidad total de códigos de boleto distintos: {total_tickets_distintos}")
print(f"Cantidad de códigos de boleto asociados a más de un pasajero: {num_tickets_multiples}")


ticket_mas_frecuente = conteo_tickets.idxmax()
frecuencia_maxima = conteo_tickets.max()
print(f"El boleto que aparece en más registros es '{ticket_mas_frecuente}', con {frecuencia_maxima} pasajeros asociados.")
```

```python

df['Tiene_Cabin'] = df['Cabin'].notna()

cabina_por_clase = df.groupby('Pclass')['Tiene_Cabin'].agg(
    Con_Cabina=lambda x: x.sum(),
    Sin_Cabina=lambda x: (~x).sum(),
    Total='count'
).reset_index()

cabina_por_clase['Clase_Nombre'] = cabina_por_clase['Pclass'].map({1: 'Primera', 2: 'Segunda', 3: 'Tercera'})
display(cabina_por_clase)

clase_max_sin_cabina = cabina_por_clase.loc[cabina_por_clase['Sin_Cabina'].idxmax()]
print(f"\nLa clase con más registros sin cabina es la Clase {clase_max_sin_cabina['Pclass']} ({clase_max_sin_cabina['Clase_Nombre']}), con {clase_max_sin_cabina['Sin_Cabina']} registros sin cabina.")
```
