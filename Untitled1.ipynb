```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np

# Cargar el archivo CSV
file_path = '/content/drive/MyDrive/'
df = pd.read_csv('clientes.csv')
df = pd.read_csv('productos_.csv')
df = pd.read_csv('ventas.csv')


print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas.")
df.head()
```

```python
#Pregunta 1
num_lineas = len(df)
num_tickets = df['id_ticket'].nunique()
total_unidades = df['cantidad'].sum()

print(f"1. Cantidad de lineas de venta: {num_lineas}")
print(f"2. Cantidad de tickets distintos: {num_tickets}")
print(f"3. Unidades vendidas: {total_unidades}")
```

```python
#Pregunta 2
tickets_por_sucursal = df[['sucursal', 'id_ticket']].drop_duplicates().groupby('sucursal').count()
tickets_por_sucursal.columns = ['num_tickets']

print("Tickets registrados por cada sucursal:")
display(tickets_por_sucursal)

sucursal_max = tickets_por_sucursal['num_tickets'].idxmax()
max_tickets = tickets_por_sucursal['num_tickets'].max()
print(f"\nLa sucursal que tuvo más tickets fue: {sucursal_max} con {max_tickets} tickets.")
```

```python
#Pregunta 3
df_tickets = df[['id_ticket', 'metodo_pago']].drop_duplicates()


df_tickets['metodo_pago'] = df_tickets['metodo_pago'].fillna('Sin registrar').replace('', 'Sin registrar')

conteo_pagos = df_tickets['metodo_pago'].value_counts()
print("Conteo de tickets por método de pago:")
display(conteo_pagos)
```

```python
#Pregunta 4
df_productos = pd.read_csv('productos_.csv')

resu_categorias = df_productos.groupby('categoria')['precio_lista'].agg(
    minimo='min',
    maximo='max',
    promedio='mean'
).reset_index()

print("Precios de lista por categoría:")
display(resumen_categorias)
```

```python
# Pregunta 6
df['importe_linea'] = df['cantidad'] * df['precio_unitario']
importe_sucursal = df.groupby('sucursal')['importe_linea'].sum().reset_index()
importe_sucursal.columns = ['sucursal', 'importe_total']

print("Importe de las ventas por cada sucursal:")
display(importe_sucursal)
```

```python
# Pregunta 7
unidades_por_ticket = df.groupby('id_ticket')['cantidad'].sum().reset_index()

def clasificar_tamano(cant):
    if cant <= 7:
        return 'Hasta 7 unidades'
    elif 8 <= cant <= 14:
        return '8 a 14 unidades'
    else:
        return '15 o más unidades'

unidades_por_ticket['tamano_compra'] = unidades_por_ticket['cantidad'].apply(clasificar_tamano)

conteo_tamano = unidades_por_ticket['tamano_compra'].value_counts()[['Hasta 7 unidades', '8 a 14 unidades', '15 o más unidades']]
print("Número de tickets por su tamaño de compra:")
display(conteo_tamano)
```

```python
# Pregunta 8
df['importe_linea'] = df['cantidad'] * df['precio_unitario']
df['fecha'] = pd.to_datetime(df['fecha'])
df['mes'] = df['fecha'].dt.to_period('M')

ventas_mes = df.groupby('mes').agg(
    num_tickets=('id_ticket', 'nunique'),
    importe_total=('importe_linea', 'sum')
).reset_index()

print("Ventas por cada mes:")
display(ventas_mes)

mes_mayor = ventas_mes.loc[ventas_mes['importe_total'].idxmax()]
print(f"\nEl que mayor mes apote tuvo fue:{mes_mayor['mes']} con un total de ${mes_mayor['importe_total']:,.2f}.")
```
